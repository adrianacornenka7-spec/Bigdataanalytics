```python
l1 = 1

v1 = np.array([4, 5, 1])

l1*v1 + l2*v2 + l3*v3

Але помістіть скаляри в список, а вектори — в ролі елементів у списку (таким чином, у вас буде два списки, один зі скалярів і один з масивів NumPy). Потім застосуйте цикл for, щоб реалізувати операцію лінійної комбінації. Ініціалізуйте вихідний вектор функцією np.zeros(). Підтвердіть, що ви отримуєте той же результат, що й в наведеному раніше вихідному коді. 


````python
# Скаляри
l1 = 1
l2 = 2
l3 = -3
#Вектори
v1 = np.array([4, 5, 1])
v2 = np.array([-4, 0, -4])
v3 = np.array([1, 3, 2])
#Лінійна комбінація
linear_combination=l1*v1 + l2*v2 + l3*v3
````

````python
scalars = [l1, l2, l3]
vectors = [v1, v2, v3]
result = np.zeros(3)
# Лінійна комбінація
for a, v in zip(scalars, vectors):
    result += a * v
````

````python
result = np.zeros(3)
# Лінійна комбінація
for a, v in zip(scalars, vectors):
    result += a * v
````

````python
print(f'Результат вихідного коду обчислення лінійної комбінації: {linear_combination}')
print(f'Результат обчислення лінійної комбінації за допомогою циклу for: {result}')
print(f"Чи однакові результати: {np.allclose(linear_combination, result)}")
````

### Завдання 3.2
 Хоча метод прокручування списків з попередньої вправи у циклі не такий ефективний, як множення матриць на вектори, він більш масштабований, ніж без циклу for. Цей факт можна дослідити, додавши в ролі елементів списків додаткові скаляри і вектори.

Що відбудеться, якщо новий доданий вектор знаходиться в $R^4$, а не в $R^3$?

````python
# Скаляри
l1 = 1
l2 = 2
l3 = -3
#Вектори
v1 = np.array([4, 5, 1])
v2 = np.array([-4, 0, -4])
v3 = np.array([1, 3, 2])
v_new = np.array([1, 3, 2, 7])

scalars = [l1, l2, l3]
vectors = [v1, v2, v3, v_new]
result = np.zeros(4)

# Лінійна комбінація
for a, v in zip(scalars, vectors):
    result += a * v
print(f'Результат обчислення лінійної комбінації: {result}')

````

А що трапиться, якщо у вас буде більше скалярів, ніж векторів?

````python
# Скаляри
l1 = 1
l2 = 2
l3 = -3
l_new = 8
#Вектори
v1 = np.array([4, 5, 1])
v2 = np.array([-4, 0, -4])
v3 = np.array([1, 3, 2])

scalars = [l1, l2, l3, l_new]
vectors = [v1, v2, v3]
result = np.zeros(3)

# Лінійна комбінація
for a, v in zip(scalars, vectors):
    result += a * v
print(f'Результат обчислення лінійної комбінації: {result}')

````

## Завдання 3.3
Визначте множину векторів, що містить один вектор (1,3). Потім створіть 100 чисел, вибраних випадковим чином із рівномірного розподілу між −4 і +4 — це ваші довільні скаляри. Помножте випадкові скаляри на вектор бази, щоб створити 100 випадкових точок у підпросторі. Нанесіть ці точки на графік.

````python
v1 = np.array([1, 3])  
scalars = np.random.uniform(-4, 4, 100)  
points_2d = np.zeros((100, 2))  
for i in range(100):
    points_2d[i] = scalars[i] * v1
fig_2d = go.Figure(data=[go.Scatter(
    x=points_2d[:,0],
    y=points_2d[:,1],
    mode='markers')])
fig_2d.show()
````

Використовуючи два вектори в $R^3$: (3,5,1) і (0,2,2). Створіть 200 скалярів. Створіть 100 точок. Відобразіть ці точки на графіку.









````python
v2_1 = np.array([3,5,1])
v2_2 = np.array([0,2,2])

scalars_2_1 = np.random.uniform(-4,4,100)
scalars_2_2 = np.random.uniform(-4,4,100)

points_3d = np.zeros((100, 3)) 
for i in range(100):
    points_3d[i] = scalars_2_1[i]*v2_1 + scalars_2_2[i]*v2_2

fig_3d = go.Figure(data=[go.Scatter3d(
    x=points_3d[:,0],
    y=points_3d[:,1],
    z=points_3d[:,2],
    mode='markers')])
fig_3d.show()
````

````python
v3_1 = 2*np.array([3,5,1])
v3_2 = np.array([3,5,1])

scalars_3_1 = np.random.uniform(-4,4,100)
scalars_3_2 = np.random.uniform(-4,4,100)

points_3d_line = np.zeros((100, 3)) 
for i in range(100):
    points_3d_line[i] = scalars_3_1[i]*v3_1 + scalars_3_2[i]*v3_2

fig_3d_line = go.Figure(data=[go.Scatter3d(
    x=points_3d_line[:,0],
    y=points_3d_line[:,1],
    z=points_3d_line[:,2],
    mode='markers')])
fig_3d_line.show()
````
