## Завдання 5.3
Розширте наявний вихідний код, щоб витягнути решту чотири блоки розміром $5 \times 5$. Потім створіть нову матрицю, у якій ці блоки будуть переставлені місцями відповідно до [Рис. 5.7].

```python
A, B = C[:5, :5], C[:5, 5:]
E, D = C[5:, :5], C[5:, 5:]

new_matrix = np.zeros_like(C)
new_matrix[:5, :5] = D      
new_matrix[:5, 5:] = E 
new_matrix[5:, :5] = B       
new_matrix[5:, 5:] = A       

vmin, vmax = C.min(), C.max()

plt.figure(figsize=(12, 5))

# початкова матриця
plt.subplot(1, 2, 1)
plt.imshow(C, cmap='viridis', interpolation='nearest', vmin=vmin, vmax=vmax)
plt.title("Початкова матриця")
for i in range(10):
    for j in range(10):
        plt.text(j, i, str(C[i, j]), ha='center', va='center', color='white', fontsize=8)
plt.axvline(x=4.5, color='white', linestyle='--', linewidth=1.5)
plt.axhline(y=4.5, color='white', linestyle='--', linewidth=1.5)

# переставлена матриця
plt.subplot(1, 2, 2)
plt.imshow(new_matrix, cmap='viridis', interpolation='nearest', vmin=vmin, vmax=vmax)
plt.title("Переставлена матриця")
for i in range(10):
    for j in range(10):
        plt.text(j, i, str(new_matrix[i, j]), ha='center', va='center', color='white', fontsize=8)
plt.axvline(x=4.5, color='white', linestyle='--', linewidth=1.5)
plt.axhline(y=4.5, color='white', linestyle='--', linewidth=1.5)

plt.show()
```

## Завдання 5.5
Створити дві матриці розміром 3 × 4 та скаляр $𝜇$. Перевірити комутативність і дистрибутивність множення скаляра на матриці, обчисливши: $𝜇(𝐴 + 𝐵)$ , $𝜇𝐴 + 𝜇𝐵$ , $𝐴𝜇 + 𝐵𝜇$. Порівняти результати і підтвердити їхню еквівалентність.

```python
np.random.seed(0)  
A = np.random.rand(3, 4)
B = np.random.rand(3, 4)
mu = np.random.rand()

res1 = mu * (A + B)
res2 = mu * A + mu * B
res3 = A * mu + B * mu

print("μ(A+B):\n",np.round(res1,2))
print("μA+μB: \n", np.round(res2,2))
print("Aμ+Bμ: \n", np.round(res3,2))
print("\nРівність 𝜇(𝐴 + 𝐵) = 𝜇𝐴 + 𝜇𝐵 = 𝐴𝜇 + 𝐵𝜇 виконується:",np.allclose(res1,res2,res3))
```

## Завдання 5.6
Реалізувати множення матриць за допомогою вкладених циклів for. Підтвердити результат за допомогою оператора @ бібліотеки NumPy.

```python
A = np.random.randint(1, 10, (2, 3))
B = np.random.randint(1, 10, (3, 2))
C = np.zeros((A.shape[0], B.shape[1]), dtype=int)

for i in range(A.shape[0]): 
    for j in range(B.shape[1]): 
        for k in range(A.shape[1]):
            C[i, j] += A[i, k] * B[k, j]

print("A =\n", A)
print("\nB =\n", B)
print("\nА ⋅ B (через цикли) =\n", C)
print("\nА ⋅ B (через @) =\n", A @ B)
```

2. Перемножте чотири матриці і транспонуйте добуток.
3. Транспонуйте кожну матрицю окремо і помножте їх, не змінюючи порядок їх слідування.
4. Транспонуйте кожну матрицю окремо і помножте їх у зворотному порядку відповідно до правила LIVE EVIL. Перевірте збіг результату кроку 2 з результатами кроку 3 і кроку 4.
5. Повторіть наведені вище кроки, але використовуйте лише квадратні матриці.

```python
L = np.random.randn(2, 6)
I = np.random.randn(6, 3)
V = np.random.randn(3, 5)
E = np.random.randn(5, 2)

p = (L @ I @ V @ E).T
print("(L @ I @ V @ E).T = \n", np.round(p,2))

q = E.T @ V.T @ I.T @ L.T
print("\n E.T @ V.T @ I.T @ L.T =\n",np.round(q,2))
print("\nРівність (L @ I @ V @ E).T = E.T @ V.T @ I.T @ L.T виконується:", np.allclose(p,q))
```

## Завдання 5.8 
У цій вправі ви напишете функцію Python, яка перевіряє матрицю на симетричність. На вході вона повинна приймати матрицю і на виході виводити булеве значення $\text{True}$, якщо матриця є симетричною, або $\text{False}$, якщо матриця є несиметричною. Майте на увазі, що малі помилки обчислювального округлення / прецизійності можуть створювати враження, що «рівні» матриці є нерівними. Отже, вам потрібно буде виконувати перевірку на рівність з деякою розумною терпимістю. Протестуйте функцію на симетричних та несиметричних матрицях.

```python
def symmetric(matrix):
    if matrix.shape[0] != matrix.shape[1]:
        return False
    return np.array_equal(matrix, matrix.T)

A = np.array([[1, 2, 3],
              [2, 5, 6],
              [3, 6, 9]])

B = np.array([[1, 0, 2],
              [3, 1, 4],
              [2, 0, 1]])

print(f"A = \n {A}\nМатриця А симетрична:{symmetric(A)}")  
print(f"\nB = \n {B}\nМатриця B симетрична:{symmetric(B)}")  
```

## Завдання 5.9
Я згадував, що існує адитивний метод створення симетричної матриці з несиметричної квадратної матриці. Опишіть цей простий метод, реалізуйте його алгоритм на Python і підтвердіть, що результат дійсно є симетричним. 

```python
def make_symmetric(A):
    return 0.5 * (A + A.T)
A = np.random.randn(3, 3)  
S = make_symmetric(A)

print("Початкова матриця A:\n", np.round(A,2))
print("\nСиметризована матриця S:\n", np.round(S,2))
print("\nМатриця S симетрична:",symmetric(S))
```

## Завдання 5.10
Повторіть другу частину вправи 3.3 (два вектори в $\mathbb{R}^3$), але використовуйте множення матриці на вектор замість множення вектора на скаляр. Тобто обчисліть $\mathbf{A}\mathbf{s}$ замість $\mu_1 \mathbf{v}_1 + \mu_2 \mathbf{v}_2$. 

```python
v1 = np.array([3, 5, 1])
v2 = np.array([0, 2, 2])

A = np.column_stack((v1, v2))  
points = []                     

for i in range(100):
    s = np.random.uniform(-4, 4, 2)  
    p = A @ s                        
    points.append(p)

points = np.array(points)
fig = go.Figure(data=[go.Scatter3d(
    x=points[:, 0],
    y=points[:, 1],
    z=points[:, 2],
    mode='markers')])
fig.show()
```

## Завдання 5.11


```python
A = np.ones((4, 4))

# діагональна матриця 
D = np.diag([1, 4, 9, 16])

# масштабування рядків
pre_mult = D @ A
print("D @ A:\n", pre_mult)

# масштабування стовпців
post_mult = A @ D
print("\nA @ D:\n", post_mult)

# квадратно-діагональна матриця (таблиця множення) ---
sqrt_D = np.diag([1, 2, 3, 4])
table_mult = sqrt_D @ A @ sqrt_D
print("\n√D @ A @ √D:\n", table_mult)
```

## Завдання 5.12
Множення матриць — це те ж саме, що і адамарове множення двох діагональних матриць. Подумайте, чому так, використовуючи папір і олівець з двома діагональними матрицями $3\times3$, а потім проілюструйте це у вихідному коді на Python.

```python
D1 = np.diag([2, 3, 4])
D2 = np.diag([5, 6, 7])

classic = D1 @ D2

adamard = D1 * D2

print("D1 @ D2 (класичне множення):\n", classic)
print("\nD1 * D2 (адамарове (поелементне) множення):\n", adamard)

print("\nКласичне та адамарове множення двох діагональних матриць однакове:", np.allclose(classic, adamard))
```
