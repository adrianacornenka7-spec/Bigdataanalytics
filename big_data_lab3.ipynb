```python
from functools import reduce
import csv
import pandas as pd
import math
```

### 1. Implement versions of map, filter, and compose using reduce.

```python
def map_reduce(func, iterable):
    return reduce(lambda acc, x: acc + [func(x)], iterable, [])

def filter_reduce(predicate, iterable):
    return reduce(lambda acc, x: acc + [x] if predicate(x) else acc, iterable, [])

def compose_reduce(*funcs):
    return reduce(lambda f, g: lambda x: f(g(x)), funcs)

print("Map using reduce:", map_reduce(lambda x: x ** 3, range(10)))
print("Filter using reduce:", filter_reduce(lambda x: x**2 > 20, range(10)))
print("Compose using reduce:", list(map(compose_reduce(math.cos, math.sqrt), range(3))))
```

### 2. Using a csv file you’ve downloaded in lab1, create a generator that will yield first 3 columns of the file. And use this generator to create a Pandas.DataFrame. Hint: use https://docs.python.org/3/library/csv.html

```python
def first_three_columns(file_name):
    with open(file_name, newline='') as f:
        reader = csv.reader(f)
        for row in reader:
            yield row[:3]  
file_name = "data/heart_2020_cleaned.csv"  
gen = first_three_columns(file_name)

df = pd.DataFrame(gen)
print(df)
```
