### How many addresses are located in each district? You can find out the answer by grouping the spatial join result based on the district name.

```python
import geopandas as gpd
districts = gpd.read_file("data/_data/Helsinki/Major_districts.gpkg")
addresses = gpd.read_file("data/_data/Helsinki/addresses.shp")
addresses = addresses.set_crs(epsg=4326)
addresses = addresses.to_crs(crs=districts.crs)
spatial_join = gpd.sjoin(addresses, districts, how="left", predicate="within").groupby("Name").size()
print(spatial_join)
```
