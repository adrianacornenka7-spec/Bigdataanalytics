```python
import pandas as pd
import geopandas as gpd
from shapely.geometry import Point
import matplotlib.pyplot as plt
```

```python
address_df = pd.read_excel("data/lviv/address.xls")
geometry = [Point(xy) for xy in zip(address_df['x'], address_df['y'])]
addresses_gpd = gpd.GeoDataFrame(address_df, geometry=geometry, crs="EPSG:4326") #набір даних адрес

microdistricts=gpd.read_file("data/lviv/microdistricts.json") #набір даних районів
kids_place_lviv=gpd.read_file("data/lviv/kids_place_lviv_geojson_2025-11-08.geojson") #дитячі майданчики
botanical_attractions_lviv=gpd.read_file("data/lviv/botanical_attractions_geojson_2025-11-11.geojson") #ботанічні пам'ятки природи
build_hist_lviv=gpd.read_file("data/lviv/build_hist_lviv_geojson_2025-11-11.geojson") #історичні пам'ятки 

#все до одніє системи координат
microdistricts = microdistricts.to_crs(epsg=3857)
kids_place = kids_place_lviv.to_crs(epsg=3857)
botanical_attractions = botanical_attractions_lviv.to_crs(epsg=3857)
build_hist = build_hist_lviv.to_crs(epsg=3857)
addresses = addresses_gpd.to_crs(epsg=3857)
```

```python
#просторове об’єднання POI з мікрорайонами
join_kids_place = kids_place.sjoin(microdistricts, predicate="within", how="inner")
join_botanical_attractions = botanical_attractions.sjoin(microdistricts, predicate="within", how="inner")
join_build_hist = build_hist.sjoin(microdistricts, predicate="within", how="inner")
```

```python
kids_count = join_kids_place.groupby("name")["geometry"].count()
botanical_count = join_botanical_attractions.groupby("name_right")["geometry"].count()
build_hist_count = join_build_hist.groupby("name")["geometry"].count()
```

```python
microdistricts["Загальна кількість POI"] = microdistricts["name"].apply(
    lambda n: (
        kids_count.get(n, 0) +
        botanical_count.get(n, 0) +
        build_hist_count.get(n, 0)))

# Створюємо колонку pop-up динамічно
microdistricts["Інформація"] = microdistricts["name"].apply(
    lambda n: (
        f"Мікрорайон: {n}<br>"
        f"Дитячі майданчики: {kids_count.get(n, 0)}<br>"
        f"Ботанічні пам’ятки: {botanical_count.get(n, 0)}<br>"
        f"Історичні будівлі: {build_hist_count.get(n, 0)}"))

m = microdistricts.explore(
    column="Загальна кількість POI", 
    cmap="viridis",                  
    legend=True,
    tooltip=["Інформація"],
    popup=["Інформація"],
    style_kwds=dict(weight=0.2, color="black"))
m
```

```python
addresses['geometry'].count()
```

```python

all_poi = pd.concat([kids_place, botanical_attractions, build_hist], ignore_index=True)
all_poi = gpd.GeoDataFrame(all_poi, geometry="geometry")
closest = addresses.sjoin_nearest(all_poi, distance_col="distance")
closest = closest.merge(
    all_poi[[all_poi.active_geometry_name]], left_on="index_right", right_index=True
)
from shapely import linestrings

closest["geometry"] = linestrings(
    closest.geometry_x.get_coordinates().to_numpy(),
    closest.geometry_y.get_coordinates().to_numpy(),
)
closest = closest.set_geometry("geometry")
```

## Висновок
- count = 30121 адрес знайшли свої найближчий POI
- mean = 301.7 → середня відстань між адресою і найближчою POI ≈ 302 м
- std = 300.6 → відстані дуже різняться: деякі дуже близько, інші далеко
- min = 0.000005 м найменша відстань до POI
- 25% = 97.2 м → 25% адрес знаходяться ближче ніж 97 м до РОІ
- 50% = 208.3 м → медіана, половина адрес ближче 208 м до РОІ
- 75% = 405.9 м → 75% адрес ближче 406 м до РОІ
- max = 2856.9 м найбільша відстань до POI

```python
"Населені пункти Львівщини в 50‑км зоні навколо Львова з координатами та середньою кількістю мешканців, які приїжджають до міста"
settlements_of_lviv_df = pd.read_excel("data/lviv/settlements_of_lviv.xlsx") 
geometry = [Point(xy) for xy in zip(settlements_of_lviv_df['homecity_longitude'], settlements_of_lviv_df['homecity_latitude'])]
settlements_of_lviv_gpd = gpd.GeoDataFrame(settlements_of_lviv_df, geometry=geometry, crs="EPSG:4326") 
#settlements_of_lviv = settlements_of_lviv_gpd.to_crs(epsg=3857) 
```

```python
settlements_of_lviv_sorted = settlements_of_lviv.sort_values(by='amount', ascending=False)
top20 = settlements_of_lviv_sorted.head(20)[['homecity', 'home_district', 'amount']]
top20 = top20.reset_index(drop=True)  
top20.index = top20.index + 1         
top20
settlements_of_lviv_sorted.head()
```

```python
import folium

top20_settlements = settlements_of_lviv_sorted.head(20)
m = folium.Map(location=[49.8397, 24.0297],zoom_start=9.4)
for i in range(len(top20_settlements)):
    row = top20_settlements.iloc[i]  # отримуємо рядок за номером
    folium.CircleMarker(
        location=[row['homecity_latitude'], row['homecity_longitude']],
        radius=5 + row['amount']/1000,
        color='blue',
        fill=True,
        fill_opacity=0.6,
        popup=f"{row['homecity']}: {row['amount']}"
    ).add_to(m)

m
```
